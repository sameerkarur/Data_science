# **Demo: ROUGE Benchmark**

This demo is designed to read a PDF file and a summary of that file, and then compute the ROUGE scores for the summary by comparing it with the original document. The ROUGE scores provide a measure of the quality of the summary.

**Note:**

*   Use the **Summary.txt** generated from the **Demo: Text_Summarizer**.



## **Steps to Perform:**


*   Step 1: Import the Necessary Libraries
*   Step 2: Read the PDF File
*   Step 3: Read the Summary File
*   Step 4: Load the ROUGE Metric

### **Step 1: Import the Necessary Libraries**

```python
# Import the libraries
import PyPDF2
from datasets import load_metric
```

## **Step 2: Read the PDF File**

*   Open the PDF file.
*   Create a **PdfReader** object for the PDF file.
*   Extract the text from each page of the PDF and concatenate it into a single string.


```python
# Read the PDF file from a URL
pdf_file = open('arxiv_impact_of_GENAI.pdf', 'rb')

# Extract the text from the PDF file
pdf_reader = PyPDF2.PdfReader('arxiv_impact_of_GENAI.pdf')
num_pages = len(pdf_reader.pages)
document_text = ""
for page in range(num_pages):
    document_text += pdf_reader.pages[page].extract_text()
```

## **Step 3: Read the Summary File**

*   Open the summary file and read its content.

```python
# Read the summary.txt file from a local path
#with open("Summary.txt", "r") as f:
    #summary = f.read()
summary_file = "Summary.txt"
with open(summary_file, "r") as f:
    summary_text = f.read()
```

## **Step 4: Load the ROUGE metric**

*   Load the ROUGE metric.
*   Compute the ROUGE scores for the summary.
*   Print the scores.



```python
# Load the ROUGE metric
metric = load_metric("rouge")

# Compute the ROUGE scores for the summary
scores = metric.compute(predictions=[summary_text], references=[document_text])

# Print the scores
print(scores)
```

```python
import pandas as pd

# Convert the ROUGE scores dictionary to a DataFrame
scores_df = pd.DataFrame.from_dict(scores, orient='index')

# Print the DataFrame
scores_df
```

### **Conclusion**

The ROUGE score output shows the precision, recall, and F-measure for different versions of the ROUGE metric (ROUGE-1, ROUGE-2, ROUGE-L, and ROUGE-Lsum). A higher score (closer to 1) indicates a better match with the reference summary. The **low**, **mid**, and **high** values represent the confidence interval of the score.
