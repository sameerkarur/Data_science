```python
def outer():
    x = 1
    def inner():
        nonlocal x
        x += 1
    inner()
    return x
print(outer())
```

**Q37.** Decorator: simple timer wrapper (use time.sleep demo).

```python
import time
def timer(fn):
    def wrapper(*a, **k):
        t0 = time.time()
        out = fn(*a, **k)
        print(f'{time.time()-t0:.4f}s')
        return out
    return wrapper
@timer
def work():
    time.sleep(0.01)
work()
```

**Q41.** Binary search iterative on sorted list.

```python
def binary_search(a, t):
    lo, hi = 0, len(a) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if a[mid] == t: return mid
        if a[mid] < t: lo = mid + 1
        else: hi = mid - 1
    return -1
print(binary_search([1,3,5,7], 5))
```

**Q42.** Count vowels in string with function.

**Q43.** Merge two sorted lists into one sorted.

```python
def merge(a, b):
    i = j = 0
    out = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]: out.append(a[i]); i += 1
        else: out.append(b[j]); j += 1
    return out + a[i:] + b[j:]
print(merge([1,3], [2,4]))
```

**Q44.** Implement `min` manually for list.

```python
def my_min(arr):
    m = arr[0]
    for x in arr[1:]:
        if x < m: m = x
    return m
print(my_min([3,1,4]))
```

**Q45.** Prime check function.

```python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0: return False
    return True
print(is_prime(17))
```

**Q46.** GCD using Euclidean algorithm.

**Q49.** Retry decorator pattern: call fn up to 3 times until success.

```python
def retry(fn, times=3):
    for _ in range(times):
        try:
            return fn()
        except Exception:
            pass
    raise RuntimeError('failed')
print(retry(lambda: 42))
```

**Q50.** Implement `chunk_list(lst, size)` → sublists.

```python
def chunk_list(lst, size):
    return [lst[i:i+size] for i in range(0, len(lst), size)]
print(chunk_list([1,2,3,4,5], 2))
```
