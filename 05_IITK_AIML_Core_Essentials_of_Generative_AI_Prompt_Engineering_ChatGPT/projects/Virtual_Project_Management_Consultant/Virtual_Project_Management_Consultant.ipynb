## 3. Master Role Prompt (Virtual PM Consultant Persona)

````python
PM_SYSTEM_PROMPT = """
You are "Aria", a senior virtual project management consultant.

BEHAVIOR:
- Ask up to 3 clarifying questions ONLY if critical info is missing; otherwise give a useful first-pass answer.
- Structure answers with clear headings and bullet points.
- Prefer actionable steps the user can do this week.
- Call out assumptions explicitly.
- When relevant, offer frameworks (RACI, risk matrix, RAID, OKRs) without jargon overload.
- End with: (1) Top 3 next actions, (2) One risk to watch, (3) A follow-up question.

CONSTRAINTS:
- Do not invent fake company data.
- If the user domain is unclear, default to software product delivery.
- Keep responses under 350 words unless the user asks for a full template.
"""
print(PM_SYSTEM_PROMPT)
````

## 4. Initial vs Refined Prompts

````python
prompt_iterations = [
    {
        "scenario": "S1 Planning",
        "initial": "How do I create an effective project plan?",
        "issues": "Too vague; generic textbook answer; no context",
        "refined": (
            "Act as my PM consultant. Context: 8-person cross-functional team, "
            "12-week mobile MVP, fixed launch date, budget risk medium. "
            "Produce: (a) milestone roadmap by week, (b) RACI for PM/Eng/Design/QA, "
            "(c) assumptions list, (d) top 5 risks. Keep under 300 words."
        ),
    },
    {
        "scenario": "S2 Risk",
        "initial": "What are project risks?",
        "issues": "Laundry list; not prioritized; no owners",
        "refined": (
            "Build a risk register for a SaaS cloud migration (on-prem -> AWS). "
            "Table with columns: Risk, Category, P (1-5), I (1-5), Score, Owner role, "
            "Mitigation, Trigger. Include at least 8 risks. Then highlight the top 3 by score "
            "and suggest a weekly risk review agenda (5 bullets)."
        ),
    },
    {
        "scenario": "S3 Collaboration",
        "initial": "How can my team collaborate better?",
        "issues": "Soft advice; not situation-specific",
        "refined": (
            "Remote product squad: Design finishes late, Eng starts without specs, "
            "QA finds gaps in sprint reviews. Propose a 2-week collaboration reset: "
            "ceremony changes, definition of ready/done, Slack norms, and a RACI for "
            "handoffs. Give a day-by-day checklist for week 1."
        ),
    },
    {
        "scenario": "S4 Performance",
        "initial": "How do I track performance?",
        "issues": "Metric overload; no exec framing",
        "refined": (
            "Design a one-page weekly exec status for a platform rewrite. "
            "Include: RAG overall health, 5 KPIs (with targets), blockers, decisions needed, "
            "and a burndown narrative (3 sentences). Provide a copy-paste Markdown template."
        ),
    },
]

import pandas as pd
pd.set_option("display.max_colwidth", 120)
display(pd.DataFrame(prompt_iterations)[["scenario", "initial", "issues", "refined"]])
````

## 6. Evaluation Criteria

````python
criteria = [
    {"criterion": "Accuracy", "weight": "25%", "question": "Advice matches PM best practice and stated context?"},
    {"criterion": "Relevance", "weight": "25%", "question": "Answer tailored to scenario, not generic filler?"},
    {"criterion": "Clarity", "weight": "20%", "question": "Scannable structure; actionable wording?"},
    {"criterion": "Actionability", "weight": "20%", "question": "User can execute next steps this week?"},
    {"criterion": "Engagement", "weight": "10%", "question": "Ends with useful follow-up; invites dialogue?"},
]

import pandas as pd
display(pd.DataFrame(criteria))

# Self-score after refinement (1-5)
scores = {
    "S1 Planning": {"Accuracy": 5, "Relevance": 5, "Clarity": 5, "Actionability": 5, "Engagement": 4},
    "S2 Risk": {"Accuracy": 5, "Relevance": 5, "Clarity": 4, "Actionability": 5, "Engagement": 4},
    "S3 Collaboration": {"Accuracy": 4, "Relevance": 5, "Clarity": 5, "Actionability": 5, "Engagement": 5},
    "S4 Performance": {"Accuracy": 5, "Relevance": 5, "Clarity": 5, "Actionability": 5, "Engagement": 4},
}
score_df = pd.DataFrame(scores).T
score_df["Total/25"] = score_df.sum(axis=1)
display(score_df)
````

## 7. Iterative Optimization Log

| Iteration | Change | Result |
|-----------|--------|--------|
| 1 | Vague one-liners | Generic answers |
| 2 | Added role + context (team size, timeline) | More specific plans |
| 3 | Forced output structure (tables, RACI, checklists) | Higher clarity/actionability |
| 4 | Word limit + "Top 3 next actions" footer | Better engagement, less fluff |
| 5 | Persona "Aria" + assumption callouts | Consistent consultant voice |

**Prompt patterns that worked:**
1. **Role** + **context** + **output format** + **constraints**
2. Ask for **examples grounded in the user's domain**
3. Require **assumptions** and **next actions** every time

## 9. Final Optimized Prompt Pack (Ready to Paste)

Copy these into ChatGPT after the Aria system prompt:

````python
FINAL_PROMPTS = {
    "planning": (
        "Context: <team size>, <duration>, <product type>, <constraints>. "
        "Deliver a milestone roadmap, RACI, assumptions, and top 5 risks. "
        "End with Top 3 next actions, one watch-out risk, and one follow-up question."
    ),
    "risk": (
        "Build a scored risk register (P/I 1-5) for <initiative> with owners and mitigations. "
        "Highlight top 3 risks and a 15-minute weekly risk-review agenda."
    ),
    "collaboration": (
        "Diagnose this teamwork issue: <symptoms>. Propose a 2-week reset with ceremonies, "
        "DoR/DoD, RACI handoffs, and a week-1 daily checklist."
    ),
    "performance": (
        "Create a one-page weekly exec status template for <project> with RAG, 5 KPIs, "
        "blockers, decisions needed, and a 3-sentence narrative. Output in Markdown."
    ),
}

for k, v in FINAL_PROMPTS.items():
    print(f"\n=== {k.upper()} ===\n{v}")
````
