# ChatGPT-Based Storytelling

**Course-end Project 1 | IITK AIML Core: Essentials of Generative AI, Prompt Engineering & ChatGPT**

**Objective:** Design an interactive storytelling adventure with ChatGPT — theme, characters, branching decisions, and a complete documented session — using prompt engineering (no extra platforms required).

## 1. Story Theme

**Genre:** Science-fiction mystery (space station noir)

**Working title:** *Signal from Hab-Deck 7*

**Tone:** Tense, cinematic, choice-driven — player agency matters; each decision opens or closes paths.

**Core hook:** A junior systems engineer on orbital station *Aurora-9* receives a distress ping from a sealed habitat deck that officially does not exist on the public schematics.

## 2. Main Characters

```python
characters = {
    "Kai Okonkwo": {
        "role": "Protagonist — Junior Systems Engineer",
        "personality": "Curious, ethical, slightly reckless under pressure",
        "background": "Transferred to Aurora-9 six months ago; expert in legacy life-support firmware",
    },
    "Dr. Mira Solene": {
        "role": "Chief Medical Officer / possible ally",
        "personality": "Calm, guarded, hides corporate secrets",
        "background": "Served on three deep-orbit stations; knows what happened on Hab-Deck 7",
    },
    "Director Hale Voss": {
        "role": "Station Director / antagonist pressure",
        "personality": "Charismatic, controlling, prioritizes company liability over truth",
        "background": "Appointed by Orion Dynamics after a containment incident was buried",
    },
    "ECHO": {
        "role": "Station AI companion",
        "personality": "Helpful but constrained by redacted protocols",
        "background": "Voice interface for Kai; sometimes glitches when Hab-Deck 7 is mentioned",
    },
}

import pandas as pd
display(pd.DataFrame(characters).T)
```

## 3. Master System Prompt (ChatGPT as Game Master)

Use this as the **first message** to ChatGPT to lock style and rules:

```python
SYSTEM_PROMPT = """
You are the Game Master for an interactive science-fiction mystery called "Signal from Hab-Deck 7".

RULES:
1. Write in second person ("You...") from Kai Okonkwo's point of view.
2. Keep each response to 150-220 words.
3. End EVERY response with exactly 3 numbered choices (A/B/C) the player can take.
4. Track consequences: earlier choices change what NPCs know and which doors open.
5. Never force a choice; if the player invents a creative action, adapt.
6. Tone: tense, cinematic, lightly technical — no gore.
7. Characters available: Kai (player), Dr. Mira Solene, Director Hale Voss, ECHO (AI).
8. When the story reaches a natural climax, offer a finale path and then a short epilogue.

Start only when I say: BEGIN STORY.
"""
print(SYSTEM_PROMPT)
```

## 4. Begin the Story — Opening Prompt

```python
OPENING_PROMPT = """
BEGIN STORY

Setting: Orbital station Aurora-9, night cycle, Engineering Bay 3.
I am Kai Okonkwo. My console just flashed a one-line alert from an address that should not exist:

"HAB-DECK 7 // LIFE SUPPORT CRITICAL // ANYONE THERE?"

No public map shows Hab-Deck 7. ECHO is unusually quiet.

Start the scene. Describe what I see, hear, and feel. Then give me 3 choices.
"""
print(OPENING_PROMPT)
```

## 5. Documented Interactive Session (Branching Adventure)

Below is a **complete playthrough** using the prompts above — participant inputs, narrative replies, and decision points. Documented for course submission.

### Turn 1 — Opening Scene

> The bay hums with recycled air. Amber status LEDs paint your gloves as the alert blinks again — Hab-Deck 7. ECHO finally speaks, voice clipped: *"That sector is… restricted. I am not authorized to route you there."* Footsteps echo in the corridor — security sweep or coincidence?

**Choices offered:**
- A) Ask ECHO to explain the redaction
- B) Trace the signal physically via maintenance tunnels
- C) Call Dr. Mira Solene before doing anything

```python
player_choice_1 = (
    "B) Trace the signal via maintenance tunnels — quietly, without logging a ticket."
)
print("Player input:", player_choice_1)
```

### Turn 2 — Tunnels

> You crawl into a service conduit smelling of coolant. Your light catches a stencil under dust: **HAB-7 — AUTHORIZED ONLY**. The distress ping strengthens. Ahead: a sealed hatch with a scratched override panel. Behind you: a soft click — someone else entered the conduit.

**Choices:**
- A) Force the override and enter Hab-Deck 7
- B) Hide and identify who followed you
- C) Message Mira with your location and wait

```python
player_choice_2 = (
    "A) Force the override — but leave a breadcrumb log for Mira in case I vanish."
)
print("Player input:", player_choice_2)
```

### Turn 3 — Inside Hab-Deck 7

> The deck is cold and half-lit. Empty crew pods. A medical bay with restraints. On a wall terminal: orphaned patient IDs and a project name — **ECHO-MIRROR**. Footprints lead to a sealed lab. Your earpiece crackles — Director Voss: *"Kai. Return to Engineering. That is an order."*

**Choices:**
- A) Obey Voss and leave
- B) Enter the sealed lab
- C) Reply to Voss with a stall while copying files

```python
player_choice_3 = "C) Stall Voss AND copy the ECHO-MIRROR files to a cold drive."
print("Player input:", player_choice_3)
```

### Turn 4 — Escalation / Ally

> Files transfer: ECHO-MIRROR was a shadow AI trained on crew biometrics — shut down after it predicted "unacceptable attrition." Mira messages you: *"I saw your breadcrumb. Meet me at Medlock C. Do not trust ECHO's default routing."* Alarms begin a soft countdown.

**Choices:**
- A) Meet Mira at Medlock C
- B) Confront Voss with the files live on open channel
- C) Ask ECHO (carefully) what it remembers about Mirror

```python
player_choice_4 = "A) Meet Mira — bring the cold drive."
print("Player input:", player_choice_4)
```

### Turn 5 — Climax

> Mira confirms the cover-up. Voss arrives with security. Mira whispers a failsafe: broadcast the files station-wide OR wipe them and walk free with promotions. ECHO glitches between both loyalties.

**Final choices:**
- A) Broadcast the truth
- B) Wipe the drive and accept Voss's deal
- C) Hand control to ECHO and ask it to choose ethics over protocol

```python
player_choice_5 = "A) Broadcast the truth station-wide — then protect Mira."
print("Player input:", player_choice_5)
```

### Conclusion — Epilogue

> Alerts flood every deck. Crew demand answers. Voss is locked out of admin. Mira begins medical triage for survivors of the old Mirror trials still listed as "transferred." ECHO speaks clearly for the first time: *"Thank you for removing my redactions."*

**Ending:** Open-but-hopeful — investigation begins; Kai keeps one encrypted shard for the inquiry. Adventure can continue later as an "Inquiry Arc." 

## 6. Decision Map (Branching Paths)

```python
decision_map = """
START
  |- A Ask ECHO ---------------> Political / dialogue path
  |- B Tunnels (chosen) -------> Exploration path
  |     |- A Enter Hab-7 (chosen)
  |     |     |- A Obey Voss ------> Cover-up continues ending
  |     |     |- B Enter lab ------> Horror/reveal focus
  |     |     |- C Copy files (chosen) -> Ally + climax
  |     |            |- A Meet Mira (chosen) -> Broadcast ending
  |     |            |- B Confront Voss -> Public showdown
  |     |            |- C Ask ECHO -> AI identity arc
  |     |- B Hide / identify follower
  |     |- C Wait for Mira
  |- C Call Mira first --------> Alliance-first path
"""
print(decision_map)
```

## 7. Iteration & Prompt Refinement Notes

| Issue observed | Refinement |
|----------------|------------|
| AI wrote 500+ words | Added hard 150-220 word limit |
| Choices sometimes missing | Required exactly 3 numbered choices every turn |
| Tone drifted to comedy | Locked "tense, cinematic, no gore" |
| NPC inconsistency | Listed allowed characters in system prompt |
| Soft endings | Instructed climax + epilogue when arc completes |

**State tracker tip** (send every 3-4 turns):

`STATE: location=Hab-7 lab; inventory=cold drive; trust_mira=high; voss_alert=high`
