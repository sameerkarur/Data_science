**Q31.** Try/except FileNotFoundError.

**Q32.** Raise ValueError with message.

```python
def parse_positive(x):
    x = float(x)
    if x <= 0: raise ValueError('must be positive')
    return x
try:
    parse_positive(-1)
except ValueError as e:
    print(e)
```

**Q33.** Custom exception class AppError.

```python
class AppError(Exception): pass
try:
    raise AppError('fail')
except AppError as e:
    print(e)
```

**Q34.** else clause in try/except.

```python
try:
    x = 1
except Exception:
    pass
else:
    print('no error', x)
```

**Q36.** Chained exception raise from.

```python
try:
    1/0
except ZeroDivisionError as e:
    err = e
try:
    raise ValueError('bad') from err
except ValueError as e:
    print(e.__cause__)
```

**Q39.** Retry reading file 3 times.

```python
def read_with_retry(path, n=3):
    for i in range(n):
        try:
            return Path(path).read_text(encoding='utf-8')
        except FileNotFoundError:
            if i == n-1: raise
print('pattern ok')
```

**Q46.** Context manager for temp write then read.

```python
from contextlib import contextmanager
@contextmanager
def temp_file(path):
    p = Path(path); p.write_text('tmp', encoding='utf-8')
    try: yield p
    finally: p.unlink(missing_ok=True)
with temp_file(REPO_ROOT/'Course_01_Python_Foundations/05_file_io_exceptions/_tmp/t.txt') as p:
    print(p.read_text())
```
