**Q5.** Demonstrate Cosine Distance ($1 - \text{CosineSimilarity}$) calculation.


```python
import numpy as np
v1, v2 = np.array([1, 0]), np.array([0, 1])
cos_sim = np.dot(v1, v2) / (np.linalg.norm(v1)*np.linalg.norm(v2))
cos_dist = 1.0 - cos_sim
print(f'Cosine Distance between orthogonal vectors: {cos_dist:.2f}')
```

**Q7.** Implement Exact Nearest Neighbors (K-NN brute force) search on a small vector array.


```python
import numpy as np
vectors = np.array([[0.1, 0.9], [0.8, 0.2], [0.5, 0.5]])
query = np.array([0.2, 0.8])
sims = np.dot(vectors, query)
best_idx = np.argmax(sims)
print('Nearest neighbor index:', best_idx, 'Similarity:', sims[best_idx])
```

**Q13.** Demonstrate how to generate deterministic character n-gram pseudo-embeddings in Python.


```python
def pseudo_embed(text: str, dim: int = 4):
    vec = [0.0] * dim
    for char in text:
        vec[ord(char) % dim] += 1.0
    return [v / (sum(vec) or 1) for v in vec]
print('Pseudo vector:', pseudo_embed('Nestle Policy'))
```
