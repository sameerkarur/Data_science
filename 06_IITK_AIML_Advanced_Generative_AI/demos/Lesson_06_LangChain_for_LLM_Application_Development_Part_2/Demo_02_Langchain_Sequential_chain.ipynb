### **Step 1: Import the necessary modules**
- Install the required LangChain and OpenAI libraries and import the necessary modules

```python
# Install modern LangChain + OpenAI clients
import sys
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
import textwrap
```

### **Step 2: Define a function to print responses from the chains**
- Define a helper function that formats and prints the outputs from the model in a readable way

```python
def display_response(response: str):
    print("\n".join(textwrap.wrap(response, width=100)))
```

### **Step 3: Initialize the chat model**


- Initialize the ChatOpenAI model that will generate responses for both chains


```python
chat_model = ChatOpenAI(
    model="gpt-4o-mini", 
    temperature=0
)
```

### **Step 4: Define the first chain**

- Define the Scenario Chain, which generates a short scenario for a new TV show episode based on a given theme


```python
scenario_template = """
You have to come up with a scenario (with a 20–50 word description)
for a new episode of the TV show "Banking Chronicles" based on the theme:

{theme_input}

ANSWER:
"""

scenario_prompt = PromptTemplate(
    input_variables=["theme_input"],
    template=scenario_template
)

scenario_chain = scenario_prompt | chat_model
```

- The output from this chain will serve as the scenario input for the next chain.

### **Step 5: Define the second chain**
- Define the Dialogue Chain, which generates a short conversation between a bank manager and a customer based on the previously generated scenario

```python
dialogue_template = """
Generate a short dialogue between a bank manager and a customer
for the TV show "Banking Chronicles" based on this scenario:

{scenario}

ANSWER:
"""

dialogue_prompt = PromptTemplate(
    input_variables=["scenario"],
    template=dialogue_template
)

dialogue_chain = dialogue_prompt | chat_model
```

- This chain depends on the scenario generated by the first chain.

### **Step 6: Define the sequential chain**
- Combine the Scenario Chain and Dialogue Chain into a sequential workflow

```python
# Step 1: Run scenario_chain and get the scenario string
scenario_step = RunnableLambda(lambda x: {"scenario": scenario_chain.invoke(x).content})

# Step 2: Feed the scenario into the dialogue_chain
sequential_chain = scenario_step | dialogue_chain
```

### **Step 7: Run the sequential chain**

*   Run the Sequential Chain with the **Investment Strategies** theme and print the resulting conversation.

```python
theme = "Investment Strategies"

result = sequential_chain.invoke({"theme_input": theme})

print("\nGenerated Episode Script:")
display_response(result.content)
```

- The model first creates a scenario for the episode and then generates a dialogue based on that scenario, demonstrating model chaining in action.
